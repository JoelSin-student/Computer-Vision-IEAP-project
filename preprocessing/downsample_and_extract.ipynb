# Downsample videos and extract images
This notebook downsamples video from 60 Hz to 10 Hz to reduce computational and storage costs, while enabling a robust second-by-second temporal analysis by averaging features over 10 frames per second.

```python
import os
import cv2
```

```python
# Directory containing the videos
directory = "."

# Naming
video_prefix = "squat_"
condition = ["pre", "per", "post"]
subjects = ["joel", "th"]

# Iterate through each subject and condition
for subject in subjects:
    for cond in condition:
        video_name = f"{video_prefix}{cond}{subject}.mp4"
        video_path = os.path.join(directory, video_name)

        # Check if the video file exists
        if not os.path.exists(video_path):
            print(f"Video file {video_name} not found.")
            continue

        # Open the video file
        cap = cv2.VideoCapture(video_path)

        # Sample the source video at 10 frames per second.
        # Setting CAP_PROP_FPS does not resample frames returned by read().
        source_fps = cap.get(cv2.CAP_PROP_FPS)
        source_frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        target_fps = 10
        if source_fps > 0:
            print(f"Sampling {video_name} at {target_fps} FPS")
        else:
            print(f"Failed to get FPS for {video_name}")
            cap.release()
            continue

        # Create a directory to save the extracted frames
        frames_dir = f"{video_prefix}{cond}{subject}_frames"
        os.makedirs(frames_dir, exist_ok=True)

        # Extract frames from the video
        frame_count = 0
        source_frame_index = 0
        while True:
            ret, frame = cap.read()
            if not ret:
                break

            # Save only source frames at the target-FPS sampling points.
            expected_sample_index = round(frame_count * source_fps / target_fps)
            if source_frame_index >= expected_sample_index:
                frame_filename = os.path.join(frames_dir, f"frame_{frame_count:04d}.jpg")
                cv2.imwrite(frame_filename, frame)
                frame_count += 1
            source_frame_index += 1

        # Release the video capture
        cap.release()
        print(f"Extracted {frame_count} frames from {video_name}")

        # Compare original and sampled frame counts
        print(f"Original frame count for {video_name}: {source_frame_count}")
        print(f"Sampled frame count for {video_name}: {frame_count}")
```
